```python
# Installing packages
import pandas as pd
```

```python
# Load plumber dataset
plumbers = pd.read_excel("Plumber_Dataset.xlsx")
```

```python
plumbers.head()
```

```python
def match_plumbers(user_lga, user_lcda, user_region):
    
    def score(row):
        if row["LGA"] == user_lga:
            return 1.0
        elif row["LCDA"] == user_lcda:
            return 0.9
        elif row["Region"] == user_region:
            return 0.7
        else:
            return 0
    
    plumbers["score"] = plumbers.apply(score, axis=1)
    
    # Filter out zero scores
    result = plumbers[plumbers["score"] > 0]
    
    # Sort
    result = result.sort_values(by="score", ascending=False)
    
    return result.head(3)
```

```python
# Example
print(match_plumbers("Ikeja", "Ojodu", "Mainland"))
```
